```python
import numpy as np
#l=1m=100cm
#R=30
#Pipe area = pi*R^2=pi*900
#Box area = l^2 = 100^2
#so prob of ball falling in the pipe = pipe area/box area

R=30
l=100
prob1=np.pi*(R-1)**2/l**2

x=np.random.uniform(0,100,1000000)
y=np.random.uniform(0,100,1000000)

distance=np.sqrt((x-50)**2+(y-50)**2)
inside=distance<=29

prob2=np.mean(inside)
print("p1", prob1)
print("p2", prob2)
```
