```python
import numpy as np
def spinner(probabilityDistribution):
    return np.random.choice(len(probabilityDistribution),p=probabilityDistribution)
prob=[0.2,0.3,0.1,0.4]
results=[]
for _ in range(100000):
    results.append(spinner(prob))
results=np.array(results)
counts=np.bincount(results)
frequencies=counts/len(results)
print(frequencies)
```

```python
import numpy as np
def spinnerdegrees(degrees):
    probdeg=[a/360 for a in degrees]
    return np.random.choice(len(degrees),p=probdeg)
angles = [45, 85, 70, 160]
results = []
for _ in range(100000):
    results.append(spinnerdegrees(angles))
results = np.array(results)
counts = np.bincount(results)
frequencies = counts / len(results)
print(frequencies)
```
